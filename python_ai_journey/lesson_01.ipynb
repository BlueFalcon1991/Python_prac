```python
patients = [
    {"name": "Rahul", "age": 52},
    {"name": "Priya", "age": 45},
    {"name": "Amit", "age": 67}
]
```

```python
for patient in patients:
    if patient["age"] >= 47:
        print(patient["name"])
```

```python
senior_patients = []

for patient in patients:
    if patient["age"] >= 47:
        senior_patients.append(patient)

print(senior_patients)
```

```python
patients = [
    {
        "id": 101,
        "name": "Rahul",
        "age": 52,
        "diagnosis": "Diabetes",
        "claim_amount": 25000
    },
    {
        "id": 102,
        "name": "Priya",
        "age": 67,
        "diagnosis": "Hypertension",
        "claim_amount": 42000
    },
    {
        "id": 103,
        "name": "Amit",
        "age": 45,
        "diagnosis": "Diabetes",
        "claim_amount": 18000
    },
    {
        "id": 104,
        "name": "Sneha",
        "age": 72,
        "diagnosis": "Heart Disease",
        "claim_amount": 85000
    },
    {
        "id": 105,
        "name": "Raj",
        "age": 34,
        "diagnosis": "Asthma",
        "claim_amount": 12000
    }
]
```

```python
for patient in patients:
    print(patient["name"])
```

```python
for patient in patients:
    if patient["age"] >= 60:
        print(patient["name"])
```

```python
for patient in patients:
    if patient["claim_amount"] >= 30000:
        print(patient["name"])
```

```python
claim_total=0
for patient in patients:
    claim_total=claim_total + patient["claim_amount"]
print(claim_total)
average_claim_amount = claim_total / len(patients)
print(average_claim_amount)
```

```python
max_patient = max(patients, key=lambda patient: patient["claim_amount"])

print(max_patient["name"])
print(max_patient["claim_amount"])
```
