```python
patients = [
    {"name": "Rahul", "age": 52},
    {"name": "Priya", "age": 45},
    {"name": "Amit", "age": 67}
]
```

```python
for patient in patients:
    if patient["age"] >= 47:
        print(patient["name"])
```

```python
senior_patients = []

for patient in patients:
    if patient["age"] >= 60:
        senior_patients.append(patient)

print(senior_patients)
```
